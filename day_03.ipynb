```python
from pyspark.sql import SparkSession
import pandas as pd
```

```python
spark = SparkSession.builder.appName("files_to_df").getOrCreate()
```

inferSchema tells Spark to automatically detect the data type of each column instead of treating everything as a string.

```python
# CSV to Pyspark DataFrame:

df1 = (
    spark
    .read
    .option("header", "false")    # in txt we dont have header So we Make False
    .option("inferSchema", "true")
    .csv("dataFiles/sample.txt")
)

# Rename columns
df1 = df1.toDF("ID", "Name", "Email", "Age", "City", "Department", "Salary", "DOB", "IsActive")
'''
or way 2
df1 = df1.toDF(*["ID", "Name", "Email", "Age", "City", "Department", "Salary", "DOB", "IsActive"])
df1.show()

'''
df1.show()
```

```python
p_df = pd.read_excel("dataFiles/sports_records.xlsx")

df2 = spark.createDataFrame(p_df)
df2.show()
```

```python
df3 = (
    spark.read
    .option("multiLine", "true")
    .json("dataFiles/fruits.json")
)

df3.show()
```
