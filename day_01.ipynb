```python
from pyspark.sql import SparkSession
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl
```

```python
spark = SparkSession.builder \
            .appName('PySpark') \
            .getOrCreate()
```

List of Tuples to spark DataFrame

```python
data1 = [
    ("Alice", 4500), 
    ("Sarvana", 5000),
    ("Chandana", 5500),
    ("Kiran", 3400)
]

columns = ["Name", "Salary"]

df1 = spark.createDataFrame(data1, columns)
df1.show()
```

```python
# Numpy - direct Spark Dataframe
```

```python
arr1 = np.array([
    [12, 23, 34],
    [45, 56, 67],
    [78, 89, 90]
])

cols = ["X", "Y", "Z"]

df3 = spark.createDataFrame(arr1, cols)
df3.show()
```

###Dictionary of lists to Pyspark DataFrame:

```python
# Method 1 : Using Pandas

dict1 = {
    "Student": ["Amol", "Swara", "Akanksha", "Vijay"],
    "Marks": [45, 55, 50, 54],
    "Class": [9, 7, 8, 5]
}

p_df = pd.DataFrame(dict1)
df4 = spark.createDataFrame(p_df)
df4.show()



# Method 2 : Zip Module


dict2 = {
    "Student": ["Amol", "Swara", "Akanksha", "Vijay"],
    "Marks": [45, 55, 50, 54],
    "Class": [9, 7, 8, 5]
}

df6 = spark.createDataFrame(zip(
    dict2["Student"],
    dict2["Marks"],
    dict2["Class"]
), schema=["Student", "Marks", "Class"])

df6.show()

```

List of Dictionaries to Pyspark DataFrame:

```python
list1 = [
    {"Student": "Amol", "Marks": 45, "Class": 9},
    {"Student": "Swara", "Marks": 55, "Class": 7},
    {"Student": "Akanksha", "Marks": 50, "Class": 8},
    {"Student": "Vijat", "Marks": 54, "Class": 5},
]

df5 = spark.createDataFrame(list1)
df5.show()
```
