````python

df = (
        spark.read.json("practice_files/employee.json")
)

df.show()
````

````python
df.write.csv("practice_files/employee.csv" ,  mode='overwrite' ,
               header=True)
````

````python
df     =    (
            spark.read \
                    .option('header',       'true')
                    .option('inferschema' , 'true')
                    .csv("Practice_files/employee.csv")
)


df.show()
````

````python
df.write.parquet('Practice_files/employee.parquet' , mode = 'overwrite' )
````

```text
In PySpark interviews, a common question is:

Why is Parquet faster than CSV?

Answer:

Columnar storage
Compressed
Stores schema
Predicate pushdown
Reads only required columns

So no header or inferSchema options are needed when reading Parquet.

````python
df     = (
    spark.read \
                
                .parquet('Practice_files/employee.parquet')
)
df.show()


''' 
unnessary because default it stored with schema

.option('header','true')
                .option('inferScheme', 'true')
                
                '''
````

````python

df.write.orc('Practice_files/employee.orc')
````

````python
df = (
            spark.read   \
                    .orc('Practice_files/employee.orc')
)


df.show()
````
