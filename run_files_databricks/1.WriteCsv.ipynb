```python
data        = [(1,'Mahveer') ,
                    (2,'Ravi') ,
                    (3,'Rajesh'),
                    (4,'Rakesh')]
schema      = ['id','name']
df          = spark.createDataFrame(data = data, schema = schema)
display(df)
```

```python
df.write.csv(path='dbfs:/Volumes/workspace/default/master/roja.csv',header= True,mode='overwrite' )
```

```python

spark.read.csv(path='dbfs:/Volumes/workspace/default/master/roja.csv',header= True).show()
```

```python
df.write.json(path='dbfs:/Volumes/workspace/default/master/roja.json', mode='overwrite' )
```

```python
spark.read.json(path='dbfs:/Volumes/workspace/default/master/roja.json').show()
spark.read.json(path='dbfs:/Volumes/workspace/default/master/*.json').show()  # load all csv Files
spark.read.json(path=['dbfs:/Volumes/workspace/default/master/roja.json','dbfs:/Volumes/workspace/default/master/roja.json']).show()   # read Multiple Path Csv Files
```
