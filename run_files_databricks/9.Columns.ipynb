```python
data        =   [     (1,'Mahveer' , 5000 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Ravi' ,4500 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'Rajesh' , 2300 ,{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',3400 ,{"hair" : "black" , "eye" : "brown"})]

from pyspark.sql.types import StructType,StructField, StringType,IntegerType,MapType



'''

# MapType(StringType(),StringType())
#  Alternate
props      =   StructType([
    StructField('hair', IntegerType()),
    StructField('eye', StringType()),
])

'''




schema      =   StructType([
    StructField('id', IntegerType()),
    StructField('name', StringType()),
    StructField('salary', IntegerType()),
    StructField('prop', MapType(StringType(),StringType()))
])

df          = spark.createDataFrame(data = data, schema = schema)
df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import lit

df1 = df.withColumn('bonus',lit(500))
df1.show()
df1.printSchema()
```

```python
from pyspark.sql.functions import col


df1.select(df1.name).show()
df1.select(df1['name']).show()
df1.select(col('name')).show()
```

```python
df1.select(df.prop.hair).show()
df1.select(df['prop.hair']).show()
df1.select(col('prop.hair')).show()
```
