````python
from pyspark.sql import SparkSession

# Create Spark session
spark = SparkSession.builder \
    .appName("TestGlobalView") \
    .getOrCreate()
````

````python
data    = [(1,'maheer',205,505),(2,'wafa',405,105)]
schema  = ['id','name','salary','bonus']

df      = spark.createDataFrame(data,schema)
df.show()


````

````python
def TotalPayment(s,b):
        return s+b


from pyspark.sql.functions import udf
from pyspark.sql.types import *

TotalPaymentUDF = udf(TotalPayment, IntegerType())
LUDF            = udf(lambda x,y : x*y, IntegerType())

df.withColumn('TotalPayments',      TotalPaymentUDF(df.salary, df.bonus)).show()
df.withColumn('TotalPayments_LUDF', LUDF(df.salary, df.bonus)).show()

````

````python
# direct we give return type
from pyspark.sql.functions import udf
from pyspark.sql.types import IntegerType

@udf(returnType=IntegerType())
def TotalPayments(s, b):
    return s + b

df1 = df.withColumn("total_payment", TotalPayments("salary", "bonus"))
df1.show()
````

````python

df1 = df.select("*", TotalPayments("salary", "bonus").alias("total_payment"))
df1.show()





````

# using udf.register

```text

1. we create createOrReplaceTempView for run sql for emps table
2.we create function for operation , transformations
3.udf.register
4. createOrReplaceTempView acts like sql table 
4.retry from emps table which we create temp
![](path)
```

````python
df.createOrReplaceTempView("emps")

def totalpay(s,b):
    return s+b
spark.udf.register(name = "totalpaySQL", f = totalpay, returnType=IntegerType())
````

````python
%sql
SELECT id,name , totalpaySQL(salary,bonus) AS totalpay FROM emps
````
