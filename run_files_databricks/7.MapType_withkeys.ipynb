```python
data = [('maha',{'hair':'black','eye':'brown'}) , ('raja' ,{'hair':'brown','eye':'black'}) ]
# schema = ['name','properties'] 
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,MapType
schema = StructType([
    StructField('name',StringType()),
    StructField('properties', MapType(StringType(),StringType()))
])
df = spark.createDataFrame(data,schema)
df.show(truncate = False)
```

```python

# get item from map Three Ways 
df.withColumn('hair', df.properties['hair']).show()
df.withColumn('hair', df.properties.getItem('hair')).show()
df.withColumn('hair', df.properties.hair).show()
```

```python
df.show()

from pyspark.sql.functions import explode
# df.withColumn('explode',explode(df.properties)).show(truncate=False)  - not working 
df.select('name','properties', explode(df.properties)).show(truncate=False)

```

```python
from pyspark.sql.functions import map_keys,map_values

df.withColumn('keys',map_keys(df.properties)).show(truncate=False)
df.withColumn('values',map_values(df.properties)).show(truncate=False)


```
