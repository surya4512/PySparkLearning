```python
from pyspark.sql.types import ArrayType,StructType,StructField,IntegerType,StringType

data = [('abc',[1,2]) ,( 'def', [3,4])]
schema = StructType([
            StructField('id', StringType()), 
            StructField('numbers', ArrayType(IntegerType()))
])
df = spark.createDataFrame(data, schema)
df.show()
df.printSchema()

```

getting First Value From Array


```python
from pyspark.sql.functions      import col
df.withColumn('firstArray', col('numbers')[0]).show()
```

Merging Two Column with array


```python
from pyspark.sql.functions import array
df = spark.createDataFrame([[1,2], [3,4]], ['num1', 'num2'])

df.withColumn('nums', array(df.num1, df.num2)).show()
```

```python
from pyspark.sql.functions import array,col
df = spark.createDataFrame([(1,'Mahveer','net','azure') ,(2,'Ravi','java','spark')], ['id', 'name' ,'pg1','pg2'])

df1 = df.withColumn('nums', array(col('pg1'),col('pg2')))   # array(df.pg1, df.pg2)
df1.show()
```

array_contains() -- sql functions is used to check if array column contains a vaule. returns null if array null, true if the array contains the value,and false if otherwise

```python
from pyspark.sql.functions import array_contains,col
df1.show()
df2 = df1.withColumn('HasJava',array_contains(col('nums'), 'java')) .select('id','name','nums','HasJava')

display(df2);
df2.printSchema()



```
