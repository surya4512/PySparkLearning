```python
data = [(1,'Maha',['dotnet','azure']),(2,'bunny',['java','laws'])]
# schema = ['id','name','skills']
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,ArrayType
schema = StructType([
    StructField('id',IntegerType()),
    StructField('name',StringType()),
    StructField('skills', ArrayType(StringType()))
])
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```

explode - to create a new row for each element in the given array column

```python
from pyspark.sql.functions import explode

df.withColumn('explode', explode(df.skills)).show()    # New column
df.withColumn('skills', explode(df.skills)).show()     # override

```

Spilt  - return Arraytype After Spiliting the string column by delimiter

```python
data = [(1,'maha','.net,azure,sql'),(2,'sony','java,laws')]

from pyspark.sql.types import StructType,StructField,IntegerType,StringType,ArrayType
schema = StructType([
    StructField('id',IntegerType()),
    StructField('name',StringType()),
    StructField('skills', StringType())
])
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()

from pyspark.sql.functions import split
df1 = df.withColumn('skils', split(df.skills,','))
df1.show()
```
