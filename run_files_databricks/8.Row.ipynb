```python
# Row - which is represented as record / row in Dataframe , one can Create a row object by using named arguments on Create a custom row like class

# Row represents a single record (row) in a DataFrame.
```

```python
from pyspark.sql import Row

row = Row(name = 'Mahveer',  salary= 2000)
print(row)
print(row.name+" : "+str(row.salary))
```

```python
from pyspark.sql import Row

row1 = Row(name = 'Mahveer',salary= 2000)
row2 = Row(name = 'Maha',salary= 3200)
data = [row1, row2]
df = spark.createDataFrame(data)
df.show()
df.printSchema()
```

```python
# Row Like Constructor
from pyspark.sql import Row


Person = Row("name", "age")
d1 = Person("Mahveer", 20)
d2 = Person("Maha", 21)
data = [d1, d2]
df = spark.createDataFrame(data)
df.show()
df.printSchema()



```

```python
from pyspark.sql import Row

data   = [Row(name = 'Mahveer' , prop = Row(hair='black', eye='black')),
          Row(name = 'Maha' , prop = Row(hair='black', eye='black'))]

df = spark.createDataFrame(data)
df.show()
df.printSchema()
```

```python
from pyspark.sql import Row

Props = Row("hair", "eye")

data = [
    Row(name="Mahveer", prop=Props("black", "black")),
    Row(name="Maha", prop=Props("brown", "black"))
]

df = spark.createDataFrame(data)
df.show(truncate=False)
df.printSchema()
```
