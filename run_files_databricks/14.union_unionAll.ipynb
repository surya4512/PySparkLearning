```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

# Large Dataset 1
df1 = spark.range(1, 10) \
           .withColumnRenamed("id", "emp_id")

# 1,1000001
# 500001, 1500001
# Large Dataset 2
df2 = spark.range(5, 10) \
           .withColumnRenamed("id", "emp_id")

print(df1.count())  # 1,000,000
print(df2.count())  # 1,000,000
```

```python
result = df1.union(df2)

print(result.count())
result.show()
```

```python
result = df1.unionAll(df2)

print(result.count())
result.show()
```

```python
result = df1.union(df2).distinct()

print(result.count())
result.orderBy(result.emp_id).show()
```
