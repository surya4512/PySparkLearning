```python
data        =   [     (1,'Mahveer' ,"M" , 5000 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Laxmi',"F" ,4500 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'seha' ,"F", 2300 ,{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',"M",3400 ,{"hair" : "black" , "eye" : "brown"}),
                    (5,'Bench',"",3400 ,{"hair" : "black" , "eye" : "brown"})]
schema      =  ['id','name','gender','salary','props']

df          = spark.createDataFrame(data,schema)
df.show(truncate=False)



```

```python
df.filter(df.gender == "M" ).show()
df.where(df.gender == 'F').show()
```

```python
df.where((df.gender == 'F') & (df.salary > 2000)).show()
df.where((df.gender == 'F') | (df.salary > 4000)).show()
```
