```python
data        =   [     (1,'Mahveer' ,"M" , 5000 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Laxmi',"F" ,4500 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'seha' ,"F", 2300 ,{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',"M",3400 ,{"hair" : "black" , "eye" : "brown"}),
                    (5,'Bench',"",3400 ,{"hair" : "black" , "eye" : "brown"})]
schema      =  ['id','name','gender','salary','props']

df          = spark.createDataFrame(data,schema)
df.show(truncate=False)


```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df1 = df.select(
            
            df.id , df.name, \
            when(condition = df.gender == "M" , value = "Male").    \
            when(condition = df.gender == "F" , value = "FeMale").    \
                otherwise("unknown").alias('gender')

        )
df1.show()
```
