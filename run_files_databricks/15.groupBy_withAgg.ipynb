```python
data        =   [     (1,'Mahveer' ,"M" , 5000,'HR' ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Laxmi',"F" ,4500 ,'SD',{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'seha' ,"F", 2300 ,'HR',{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',"M",3400 ,'SD',{"hair" : "black" , "eye" : "brown"}),
                    (5,'Bench',"",3400 ,'DF',{"hair" : "black" , "eye" : "brown"})]
schema      =  ['id','name','gender','salary','department','props']

df          = spark.createDataFrame(data,schema)
df.show(truncate=False)


df1         = df.select('id','name','gender','salary','department')
df1.show()
```

groupBy - function used to collect the indentical data into groups on funtions on the group data


```python

df1.groupBy('department').count().show()

```

```python

df1.groupBy('department').min('salary').show()


df1.groupBy('department').max('salary').show()

```

```python
df1.groupBy('department').avg('salary').show()
```

GroupBy function can use only one Aggregate Function 

GroupBy with Aggregate Function -- we can use Multiple Aggreagate Function at Onces

```python
from pyspark.sql.functions import count
df.groupBy('department').agg(count('*').alias("count")).show()
```

```python
from pyspark.sql.functions import count,max,min
df.groupBy('department').agg(count('*').alias("count") , \
                             max('salary').alias("max_salary") , \
                             min('salary').alias("min_salary")).show()
```
