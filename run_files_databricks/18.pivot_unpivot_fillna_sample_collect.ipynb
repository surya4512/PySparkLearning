```python
from pyspark.sql import SparkSession

# Create Spark Session
spark = SparkSession.builder \
    .appName("EmployeeData") \
    .getOrCreate()

# Sample Data
data = [
    (1, "maheer",  "male",   "IT"),
    (2, "wafa",    "male",   "IT"),
    (3, "asi",     "female", "HR"),
    (4, "annu",    "female", "IT"),
    (5, "shakti",  "female", "IT"),
    (6, "pradeep", "male",   "HR"),
    (7, "sarfaraj","male",   "HR"),
    (8, "ayesha",  "female", "IT")
]

# Column Names
columns = ["id", "name", "gender", "dep"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show DataFrame
df.show()
```

```python
df.groupBy('dep','gender').count().show()
```

```python
df.groupBy('dep').pivot('gender').count().show()
```

```python
df.groupBy('dep').pivot('gender',['male','female']).count().show()    
# ['male','female'] this for Which element we need to show in gender
```

###UnPivot


```python
df1 = df.groupBy('dep').pivot('gender',['male','female']).count()
df1.show()
```

```python
from pyspark.sql.functions import expr
# expr - run function using text 
# for doing unpivot we dont have any expr for making so use expr and stack to do that.



df1.select('dep', expr("stack(2,'male',male,'female',female) as (gender,count)")).show()
```

```python
data = [(1,'maheer',2000),(2,'wafa',3000)]
schema = ['id','name','salary']

df = spark.createDataFrame(data, schema)

df.show()

from pyspark.sql.functions import upper

def convertToUpper(df):
    return df.withColumn('name_capital', upper(df.name))

def doubleTheSalary(df):
    return df.withColumn('salary_double', df.salary * 2)

df1 = df.transform(convertToUpper) \
        .transform(doubleTheSalary)

df1.show()
```

# using transform function

```python
data = [
    (1, 'maheer', ['azure', 'dotnt']),
    (2, 'wafa', ['aws', 'java'])
]

schema = ['id', 'name', 'skills']

df = spark.createDataFrame(data, schema)

df.show(truncate=False)
df.printSchema()

from pyspark.sql.functions import transform, upper

# Using lambda function
df.select(
    'id',
    'name',
    transform('skills', lambda x: upper(x)).alias('skills')
).show(truncate=False)

# Using reusable function
def convertToUpper1(x):
    return upper(x)

df.select(
    transform('skills', convertToUpper1).alias('skills')
).show(truncate=False)
```
