```python
data        =   [     (1,'Mahveer' ,"M" , 5000 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Laxmi',"F" ,4500 ,{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'seha' ,"F", 2300 ,{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',"M",3400 ,{"hair" : "black" , "eye" : "brown"}),
                    (5,'Bench',"",3400 ,{"hair" : "black" , "eye" : "brown"})]
schema      =  ['id','name','gender','salary','props']

df          = spark.createDataFrame(data,schema)
df.show(truncate=False)


```

alias

```python
df.select(df.id,df.name,df.gender,df.salary,df.props["hair"]).show()   # without alias
df.select(df.id,df.name,df.gender.alias('gender_type'),df.salary,df.props["hair"]).show() # with alias at gender column name
```

asc, desc


```python
df.sort(df.name.asc()).show()
df.sort(df.name.desc()).show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df1 = df.select("id","gender",df.salary)
df1.printSchema()

# with Cast
df2 = df.select("id","gender",df.salary.cast('double'))
df2.printSchema()
```

Like Function same as SQL

```python
df1 = df.filter(df.name.like('M%') | df.name.like('L%'))
df1.show()

df1 = df.filter(df.name.like('M%') )
df1.show()

df.filter(df.salary >= 5000).show()
```
