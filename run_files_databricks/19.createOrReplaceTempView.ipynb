```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('spark_text').getOrCreate()
```

```python
data = [(1, 'maheer', 2000),
        (2, 'asi', 3000)]

schema = ['id', 'name', 'salary']

df = spark.createDataFrame(data, schema)

# Create Temporary View
df.createOrReplaceTempView('employees')

# Query using Spark SQL
df1 = spark.sql("SELECT * FROM employees")

df1.show()
```

```python
# Query using Spark SQL
df1 = spark.sql("SELECT id , UPPER(name) FROM employees")

df1.show()
```

```python
spark.sql("SELECT * FROM employees").show()    # only works when we create Tempview
```
