```text
partitionBy()

- Used while writing DataFrames.
- Creates separate folders based on column values.
- Improves query performance through partition pruning.
- Commonly used with Parquet and ORC files.

Single Partition:
df.write.partitionBy("dep")

Multiple Partitions:
df.write.partitionBy("dep", "gender")
```

````python
data = [
    (1, 'maheer', 'male', 'IT'),
    (2, 'wafa', 'male', 'HR'),
    (3, 'asi', 'female', 'IT')
]

schema = ['id', 'name', 'gender', 'dep']

df = spark.createDataFrame(data, schema)




````
