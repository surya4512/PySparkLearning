```python
data        = [(1,'Mahveer') ,
                    (2,'Ravi') ,
                    (3,'Rajesh'),
                    (4,'Rakesh')]
schema      = ['id','name']
df          = spark.createDataFrame(data = data, schema = schema)
df.show()
```

```python
from pyspark.sql.functions import lit
df1         = df.withColumn('master',  lit('MahveerMahveerMahveerMahveerMahveer'))
df1.show()
```

```python
df1.show(n = 2 , truncate=False , vertical=False)
# n = Number of Rows
# truncate = False, to show all columns
# vertical = False, vertical view
#  
```

```python
df1.show(n = 2 , truncate=False , vertical=True)
```

```python
df1.show(n = 2 , truncate=True , vertical=True)
```
