UnionByName - let you to merge/union two DatafRame with a differeent number of columns (diff schema) by Passing allow missing columns with the Value true.

```python

df1 = spark.createDataFrame([
    (1, "John",'male')
], ["id", "name" , "gender"])

df2 = spark.createDataFrame([
    ( 2 ,"Mary",2000)
], ["id" ,"name",  "salary"])

''' 
r1 = df1.union(df2)
r1.show()
'''

result = df1.unionByName(allowMissingColumns=True, other = df2)
result.show()


```
