```python
data        =   [     (1,'Mahveer' , 5000 ) ,
                    (2,'Ravi' ,4500 ) ,
                    (3,'Rajesh' , 2300 ),
                    (4,'Rakesh',3400 )]
schema      =   ['id','name' , 'salary']

df          = spark.createDataFrame(data = data, schema = schema)
df.show()

```

col  - we have load column in this dataframe 
cast - Change Datatype

```python
# withColumn  --- overwrite -withColumn
#             --- create new column 
```

```python
from pyspark.sql.functions import col

df1 = df.withColumn("new_salary" , col("salary"))
df1.show()
df1.printSchema()
```

```python
from pyspark.sql.functions import col

df2 = df.withColumn("new_salary" , col("salary").cast("Integer"))
df2.show()
df2.printSchema()
```

```python
from pyspark.sql.functions import col,lit

df1 = df.withColumn("new_salary_modify" , col("salary") * 2).withColumn("lit" , lit('Mylit'))
df1.show()
df1.printSchema()
```



withColumnRenamed



```python
from pyspark.sql.functions import col

df1 = df.withColumn("new_salary" , col("salary")*2)


df2 = df1.withColumnRenamed("new_salary" , 'RenamedSalary')
df2.show()
df2.printSchema()
```
