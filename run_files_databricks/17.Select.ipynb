select() is used to choose specific columns from a DataFrame and optionally create derived columns.

select() is a transformation used to project specific columns from a DataFrame, similar to the SQL SELECT clause, and it can also create derived columns using expressions.

```python
data        =   [     (1,'Mahveer' ,"M" , 5000,'HR' ,{"hair" : "black" , "eye" : "brown"}) ,
                    (2,'Laxmi',"F" ,4500 ,'SD',{"hair" : "black" , "eye" : "brown"}) ,
                    (3,'seha' ,"F", 2300 ,'HR',{"hair" : "black" , "eye" : "brown"}),
                    (4,'Rakesh',"M",3400 ,'SD',{"hair" : "black" , "eye" : "brown"}),
                    (5,'Bench',"",3400 ,'DF',{"hair" : "black" , "eye" : "brown"})]
schema      =  ['id','name','gender','salary','department','props']

df          = spark.createDataFrame(data,schema)
df.show(truncate=False)


df1         = df.select('id','name','gender','salary','department')
df1.show()
```

```python

df.select(df.id,df.name).show()

df.select(df['id'],df['name']).show()
```

```python
from pyspark.sql.functions import col
df.select(col('id'),col('name')).show()
df.select(['id','name']).show()
df.select(df.id,df.name).show()
df.select('*').show()
```

```python
df.select([col for col in df.columns]).show()
```
