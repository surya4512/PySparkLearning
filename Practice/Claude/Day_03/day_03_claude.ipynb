```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, broadcast

spark = SparkSession.builder \
    .appName("Day3-Joins") \
    .getOrCreate()

# Load both tables
customers = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("customers.csv")

orders = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("orders.csv")

customers.printSchema()
orders.printSchema()
customers.show(5)
orders.show(5)
```

```python
# Inner join customers with orders on customer_id. Show name, city, product, amount.
```

```python
# Inner join customers with orders on customer_id. Show name, city, product, amount.
# Inner join and select specific columns
customers.join(orders, on="customer_id", how="inner") \
         .select("name", "city", "product", "amount") \
         .show()
```

```python
from pyspark.sql.functions import col

customers.join(orders, on="customer_id", how="left") \
         .filter(col("product").isNull()) \
         .select("name", "city", "product", "amount") \
         .show()


''' 
Not null 

# Left join to find all customers who have NEVER placed an order (use filter after join).
customers.join(orders, on="customer_id", how="left") \
         .select("name", "city", "product", "amount") \
         .show()
         
         
'''
```

```python

## Anti-join: find orders with no matching customer record.


# Find orders that do not have a matching customer_id in the customers table
orphaned_orders = orders.join(customers, on="customer_id", how="left_anti")

orphaned_orders.show()




```
